## 1. Carregando os dados

Vamos trabalhar com os dois datasets do módulo: um pequeno para consolidar a base e outro maior para treinar um fluxo mais realista.


```python
import pandas as pd

df_escola = pd.read_csv('../../datasets/escolhidos_ava_2/EquipeA_escola.csv')
df_cidades = pd.read_csv('../../datasets/escolhidos_ava_2/EquipeE_cidades.csv')
```

```python
# Sempre vale começar olhando as primeiras linhas.
# Isso ajuda a reconhecer o que cada coluna representa.
display(df_escola.head())
display(df_cidades.head(3))
```

---
## 2. Miniestudo 1: desempenho escolar

Como o dataset é pequeno, conseguimos percorrer todo o ciclo de leitura, transformação e síntese com poucas linhas.


```python
# Criamos uma cópia para evitar alterar o DataFrame original por engano.
escola = df_escola.copy()

# A coluna 'situacao' será usada em várias análises posteriores.
escola['situacao'] = escola.apply(
    lambda linha: 'Aprovado' if linha['nota'] >= 7 and linha['frequencia'] >= 75 else 'Reprovado',
    axis=1
)

# Agora já conseguimos enxergar o resultado da regra de negócio.
escola
```

```python
# Resumo rápido do conjunto.
resumo_escola = {
    'media_nota': escola['nota'].mean(),
    'media_frequencia': escola['frequencia'].mean(),
    'quantidade_aprovados': (escola['situacao'] == 'Aprovado').sum(),
    'quantidade_reprovados': (escola['situacao'] == 'Reprovado').sum()
}
resumo_escola
```

```python
# Também podemos ordenar do melhor para o pior desempenho.
ranking_escola = escola.sort_values(['nota', 'frequencia'], ascending=False)
ranking_escola
```

---
## 3. Miniestudo 2: cidades brasileiras

O segundo dataset é grande. Por isso, um bom caminho é selecionar primeiro apenas as colunas relevantes para o problema.


```python
# Escolhemos variáveis ligadas a população, renda, mobilidade e turismo.
colunas = [
    'CITY', 'STATE', 'CAPITAL', 'ESTIMATED_POP', 'GDP_CAPITA', 'IDHM',
    'HOTELS', 'BEDS', 'Cars', 'Motorcycles', 'RURAL_URBAN'
]
cidades = df_cidades[colunas].copy()
cidades.head()
```

```python
# Renomear colunas pode melhorar muito a leitura de quem está começando.
cidades = cidades.rename(columns={
    'CITY': 'cidade',
    'STATE': 'uf',
    'CAPITAL': 'capital',
    'ESTIMATED_POP': 'populacao_estimada',
    'GDP_CAPITA': 'pib_per_capita',
    'IDHM': 'idhm',
    'HOTELS': 'hoteis',
    'BEDS': 'leitos',
    'Cars': 'carros',
    'Motorcycles': 'motos',
    'RURAL_URBAN': 'perfil_territorial'
})
cidades.head()
```

```python
# Primeiro diagnóstico: há valores ausentes?
cidades.isna().sum()
```

```python
# Criando indicadores derivados para facilitar perguntas analíticas.
cidades['tem_rede_hoteleira'] = cidades['hoteis'] > 0
cidades['frota_total'] = cidades['carros'] + cidades['motos']
cidades['faixa_idhm'] = pd.cut(
    cidades['idhm'],
    bins=[0, 0.6, 0.7, 0.8, 1.0],
    labels=['baixo', 'medio', 'alto', 'muito_alto']
)
cidades[['cidade', 'uf', 'idhm', 'faixa_idhm', 'frota_total']].head()
```

```python
# Exemplo de pergunta 1: quais são as cidades mais populosas?
mais_populosas = cidades.nlargest(10, 'populacao_estimada')[['cidade', 'uf', 'populacao_estimada']]
mais_populosas
```

```python
# Exemplo de pergunta 2: quais estados têm maior média de IDHM?
idhm_por_uf = (
    cidades.groupby('uf')['idhm']
    .mean()
    .sort_values(ascending=False)
    .round(4)
)
idhm_por_uf.head(10)
```

```python
# Exemplo de pergunta 3: capitais têm, em média, IDHM maior?
idhm_capital = cidades.groupby('capital')['idhm'].mean().round(4)
idhm_capital
```

```python
# Exemplo de pergunta 4: quais cidades se destacam em turismo?
# Aqui usamos hotéis e leitos como aproximação simples.
turismo = cidades[cidades['hoteis'] > 20].sort_values(['hoteis', 'leitos'], ascending=False)
turismo[['cidade', 'uf', 'hoteis', 'leitos']].head(15)
```
