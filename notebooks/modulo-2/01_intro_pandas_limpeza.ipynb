**Objetivo:** introduzir leitura, limpeza e manipulação básica de dados com Pandas.  
**Datasets usados:** `EquipeA_escola.csv` e `EquipeE_cidades.csv`.


---
## 1. Preparação do ambiente

Neste módulo, vamos trabalhar com tabelas. O Pandas organiza os dados em estruturas chamadas **Series** e **DataFrames**.


```python
import pandas as pd

# Caminhos relativos ao notebook
CAMINHO_ESCOLA = '../../datasets/escolhidos_ava_2/EquipeA_escola.csv'
CAMINHO_CIDADES = '../../datasets/escolhidos_ava_2/EquipeE_cidades.csv'
```

---
## 2. Primeiro contato com um DataFrame

Começamos com o dataset da escola, que é pequeno e ideal para entender os comandos básicos.


```python
# Leitura do CSV mais simples
df_escola = pd.read_csv(CAMINHO_ESCOLA)
df_escola
```

```python
# Observando dimensões, tipos e estatísticas iniciais
print('Linhas e colunas:', df_escola.shape)
print('\nTipos de dados:')
print(df_escola.dtypes)
print('\nResumo numérico:')
display(df_escola.describe())
```

---
## 3. Seleção de colunas e filtros

Uma das tarefas mais comuns em análise de dados é selecionar partes da tabela.


```python
# Selecionando colunas específicas
df_escola[['nome', 'nota']]
```

```python
# Filtrando alunos com nota maior ou igual a 8
df_escola[df_escola['nota'] >= 8]
```

```python
# Filtrando alunos aprovados considerando nota e frequência
aprovados = df_escola[(df_escola['nota'] >= 7) & (df_escola['frequencia'] >= 75)]
aprovados
```

---
## 4. Criando novas colunas

Transformar dados também envolve criar colunas com regras simples.


```python
# Classificando a situação de cada aluno
df_escola['situacao'] = df_escola.apply(
    lambda linha: 'Aprovado' if linha['nota'] >= 7 and linha['frequencia'] >= 75 else 'Reprovado',
    axis=1
)
df_escola
```

```python
# Estatísticas por situação
df_escola.groupby('situacao')[['nota', 'frequencia']].mean().round(2)
```

---
## 5. Limpeza de dados

Mesmo quando um dataset parece pronto, vale conferir duplicatas, valores ausentes e nomes de colunas.


```python
# Verificações simples de qualidade dos dados
print('Valores ausentes por coluna:')
print(df_escola.isna().sum())
print('\nLinhas duplicadas:', df_escola.duplicated().sum())
```

```python
# Exemplo didático de limpeza: padronizando nomes de colunas
df_escola_limpo = df_escola.rename(columns={
    'nome': 'aluno',
    'nota': 'nota_final',
    'frequencia': 'frequencia_percentual'
})
df_escola_limpo.head()
```

---
## 6. Avançando para um dataset maior

Agora usamos `EquipeE_cidades.csv`, que possui muitas colunas e permite análises mais próximas de um contexto real.


```python
# Lendo apenas parte do dataset para um primeiro reconhecimento
df_cidades = pd.read_csv(CAMINHO_CIDADES)
print('Formato:', df_cidades.shape)
display(df_cidades.head())
```

```python
# Selecionando um subconjunto mais amigável para começar a análise
colunas_principais = [
    'CITY', 'STATE', 'CAPITAL', 'ESTIMATED_POP', 'GDP_CAPITA',
    'IDHM', 'HOTELS', 'BEDS', 'Cars', 'Motorcycles', 'RURAL_URBAN'
]
cidades_base = df_cidades[colunas_principais].copy()
cidades_base.head()
```

```python
# Investigando valores ausentes e tipos
print(cidades_base.isna().sum())
print('\nTipos de dados:')
print(cidades_base.dtypes)
```

```python
# Exemplos de filtros úteis
capitais = cidades_base[cidades_base['CAPITAL'] == 1]
cidades_turisticas = cidades_base[cidades_base['HOTELS'] > 20]
display(capitais[['CITY', 'STATE', 'IDHM', 'GDP_CAPITA']].head())
display(cidades_turisticas[['CITY', 'STATE', 'HOTELS', 'BEDS']].head(10))
```

```python
# Ordenando para responder perguntas simples
top_gdp = cidades_base.nlargest(10, 'GDP_CAPITA')[['CITY', 'STATE', 'GDP_CAPITA']]
top_pop = cidades_base.nlargest(10, 'ESTIMATED_POP')[['CITY', 'STATE', 'ESTIMATED_POP']]
display(top_gdp)
display(top_pop)
```

```python
# Resumo por estado: média de IDHM e PIB per capita
resumo_estado = (
    cidades_base.groupby('STATE')[['IDHM', 'GDP_CAPITA']]
    .mean()
    .sort_values('IDHM', ascending=False)
    .round(3)
)
resumo_estado.head(10)
```
